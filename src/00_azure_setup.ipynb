```python
# 00 - Azure HydroFlow Environment Setup

# Configure Unity Catalog objects required by the Azure-backed HydroFlow pipeline.
```

```python
spark.sql("""
DESCRIBE STORAGE CREDENTIAL hydroflow_adls_credential
""").show(truncate=False)
```

```python
storage_account = "sthydroflowlab1"
container = "hydroflow"

landing_url = (
    f"abfss://{container}@{storage_account}"
    ".dfs.core.windows.net/landing/"
)

managed_url = (
    f"abfss://{container}@{storage_account}"
    ".dfs.core.windows.net/managed/"
)
```

```python
spark.sql(f"""
CREATE EXTERNAL LOCATION IF NOT EXISTS hydroflow_managed
URL '{managed_url}'
WITH (STORAGE CREDENTIAL hydroflow_adls_credential)
""")
```

```python
spark.sql(f"""
CREATE CATALOG IF NOT EXISTS hydroflow
MANAGED LOCATION '{managed_url}catalogs/hydroflow/'
""")
```

```python
for schema_name in ["raw", "bronze", "silver", "gold"]:
    spark.sql(
        f"CREATE SCHEMA IF NOT EXISTS hydroflow.{schema_name}"
    )
```

```python
spark.sql(f"""
CREATE EXTERNAL VOLUME IF NOT EXISTS hydroflow.raw.landing
LOCATION '{landing_url}'
""")
```

```python
spark.sql("""
CREATE VOLUME IF NOT EXISTS hydroflow.bronze.checkpoints
""")

spark.sql("""
CREATE VOLUME IF NOT EXISTS hydroflow.bronze.schemas
""")

spark.sql("""
CREATE VOLUME IF NOT EXISTS hydroflow.silver.checkpoints
""")
```

```python
print("Azure HydroFlow Unity Catalog environment configured.")
```
